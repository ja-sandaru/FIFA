```python
import numpy as np
import pandas as pd
```

```python
data = pd.read_csv("results.csv")
data
```

```python
x = data[["date","home_team","away_team","Result_team1"]]
x.tail()
```

```python
data1 = pd.read_csv("new filed.csv")
data1
```

```python
def get_recent(row):
    input_date = pd.to_datetime(row["Match_Date"])
    fi_team = row["Team1"]
    # fi_team2 = row["Team2"]

    # filter date 
    date_fill = x[pd.to_datetime(x["date"]) < input_date ]
    
    # filter for the home_team to fil team
    cond1 = (
        date_fill["home_team"] == fi_team

    )
    # filter for the away_team to fil team
    cond2 =(
        date_fill["away_team"] == fi_team
    )

    recent_form = date_fill[cond1 | cond2].tail()

    points = 0
    
    for r,i in recent_form.iterrows():
        if i["home_team"] == fi_team:

            if i["Result_team1"] == "Win":
                points = points + 1
            elif i["Result_team1"] == "Draw":
                points = points + 0.5 

        else:
            if i["Result_team1"] == "Lose":
                points = points +1
            elif i["Result_team1"] == "Draw":
                points = points + 0.5

    
    
    
    return points            

    
    
    


data1["Team1_Recent_form"] = data1.apply(
    get_recent,
    axis=1
)


```
